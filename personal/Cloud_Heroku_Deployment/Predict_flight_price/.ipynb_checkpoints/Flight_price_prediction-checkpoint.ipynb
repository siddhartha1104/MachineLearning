# Feature Engineering

```python
 df.head(3)
```

# Date of Journey

```python
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'])
df['Day_of_Journey'] = (df['Date_of_Journey']).dt.day
df['Month_of_Journey'] = (df['Date_of_Journey']).dt.month
df.drop(['Date_of_Journey'], axis=1, inplace= True)
```

```python
df.head(3)
```

# Dep time

```python
df['Dep_Hour'] = pd.to_datetime(df['Dep_Time']).dt.hour
df['Dep_min'] = pd.to_datetime(df['Dep_Time']).dt.minute
df.drop(['Dep_Time'], axis=1, inplace=True)
```

```python
df.head(3)
```

# Arrival Time

```python
df['Arrival_hour'] = pd.to_datetime(df['Arrival_Time']).dt.hour
df['Arrival_min'] = pd.to_datetime(df['Arrival_Time']).dt.minute
df.drop(['Arrival_Time'], axis=1, inplace = True)
```

```python
df.head(3)
```

# Duration

```python
duration = df['Duration'].str.split(' ', expand= True)
duration[1].fillna('00m', inplace= True)
```

```python
df.head(3)
```

```python
df['duration_hr']= duration[0].apply(lambda x: x[:-1])
df['duratoin_min'] = duration[1].apply(lambda x: x[:-1])
```

```python
df.head(3)
```

```python
df.dtypes
```

```python
df['duration_hr'] = df['duration_hr'].astype(int)
df['duratoin_min'] = df['duratoin_min'].astype(int)
```

```python
df.dtypes
```

```python
df.drop(['Duration'], axis = 1, inplace=True)
```

# Dividing our data into dependent and independent 

```python
x = df.drop(['Route','Additional_Info','Price'], axis = 1)
y = df['Price']
```

```python
print('train_shape',df.shape)
print('test_shape',df_test.shape)
```

# Model Building and Hyperparameter Tuning

```python
# Spliting the dataset

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state=50)
```

# Model ==> ExtraTreesRegressor

```python
from sklearn.ensemble import ExtraTreesRegressor
ET_Model = ExtraTreesRegressor(n_estimators=120)
ET_Model.fit(x_train,y_train)

```

```python
y_predict = ET_Model.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test, y_predict)
```

# Model ==> RandomForestRegressor

```python
from sklearn.ensemble import RandomForestRegressor
RF_Model = RandomForestRegressor()
RF_Model.fit(x_train,y_train)
```

```python
y_predict = RF_Model.predict(x_test)
r2_score(y_test, y_predict)
```
